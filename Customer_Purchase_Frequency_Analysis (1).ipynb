**Data** **Collection**

```python

import csv

with open("customer_purchases.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["CustomerID", "Date", "Amount"])

    n = int(input("Enter number of purchases: "))

    for i in range(n):
        print("\nEnter details for purchase", i+1)
        customer_id = input("Customer ID: ")
        date = input("Purchase Date (DD-MM-YYYY): ")
        amount = input("Purchase Amount: ")

        writer.writerow([customer_id, date, amount])

print("\nData collection completed and saved to CSV file")
```

**Data** **Preprocessing**

```python

import csv

cleaned_data = []

with open("customer_purchases.csv", "r") as file:
    reader = csv.reader(file)

    header = next(reader)

    for row in reader:
        customer_id = row[0]
        date = row[1]
        amount = row[2]

        cleaned_data.append([customer_id, date, amount])

print("Preprocessed Data:\n")
for data in cleaned_data:
    print(data)
```

**Data** **Processing**

```python

import csv

purchase_count = {}

with open("customer_purchases.csv", "r") as file:
    reader = csv.reader(file)
    header = next(reader)

    for row in reader:
        customer_id = row[0]

        if customer_id in purchase_count:
            purchase_count[customer_id] += 1
        else:
            purchase_count[customer_id] = 1


print("Customer Purchase Count:\n")
for customer, count in purchase_count.items():
    print(customer, ":", count)
```

**Purchase** **Frequency** **Analysis**

```python

purchase_count = {
    "C001": 3,
    "C002": 2,
    "C003": 1
}

print("Purchase Frequency Analysis:\n")

for customer, count in purchase_count.items():
    if count >= 3:
        frequency = "Frequent Customer"
    elif count == 2:
        frequency = "Moderate Customer"
    else:
        frequency = "Low Frequency Customer"

    print(customer, "->", count, "purchases ->", frequency)
```

**Result** **Display**

```python

analysis_result = {
    "C001": "Frequent Customer",
    "C002": "Moderate Customer",
    "C003": "Low Frequency Customer"
}

print("FINAL RESULT – CUSTOMER PURCHASE FREQUENCY\n")
print("Customer ID\tPurchase Category")
print("--------------------------------")

for customer, category in analysis_result.items():
    print(customer, "\t\t", category)
```
